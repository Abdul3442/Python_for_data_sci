# Introduction to pandas
Welcome to Pandas! If you are learning data science or data analysis in Python, Pandas is one of the most important tools you will ever use. Lets discuss in detail!

> Pandas is a Python library used for working with data.
> Think of Pandas as an Excel spreadsheet inside Python. It helps you clean, change, analyze, and look at your data easily using code.

## Why do we use Pandas?

> It handles tables of data (like rows and columns) very quickly.
>
> It can open many file types like CSV, Excel, and JSON.
>
> It helps fix missing or messy data with just one line of code.

## The Two Main Building Blocks of Pandas

Pandas has two main types of data structures: Series and DataFrame.

1. Series (One-Dimensional)

A Series is like a single column in a table (or a list of items with labels).

```python
import pandas as pd

# A simple list of fruits
fruits = ["Apple", "Banana", "Cherry"]

# Convert it into a Pandas Series
my_series = pd.Series(fruits)

print(my_series)
```

2. DataFrame (Two-Dimensional)

A DataFrame is a full table with rows and columns. This is what you will use 99% of the time!

```python
import pandas as pd

# Create a dictionary with data
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Paris"]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

print(df)
```

## Easy Operations on a DataFrame

1. Let us look at how to do simple things with our DataFrame **df.**
> 
2. Viewing the Top Rows
If your table has thousands of rows, you only want to see the first few. Use. **head():**

```python
print(df.head(2)) # Shows the first 2 rows
```

## Looking at Column Names and Data Types

```python
print(df.columns) # Shows column names
print(df.info())  # Shows data types and missing values
```

## Selecting a Single Column
If you only want to see the "Name" column:

```python
print(df["Name"])
```

## Adding a New Column

Let us add a "Salary" column to our table:

```python
df["Salary"] = [50000, 60000, 75000]
print(df)
```
